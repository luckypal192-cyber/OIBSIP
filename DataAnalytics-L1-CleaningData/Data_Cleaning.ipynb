```python
import pandas as pd
import numpy as np

# Load the raw dataset
df = pd.read_csv("dataset/retail_store_sales.csv")

# Display basic information
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Data Quality Report - Numerical Value Range Check

print("=== NUMERICAL COLUMNS SUMMARY ===")
display(df.describe(include="all").T)
```

```python
# Outlier Detection using IQR Method

numeric_columns = df.select_dtypes(include=np.number).columns

outlier_report = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_report.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_report)

display(outlier_report)
```

## Data Cleaning Strategy

The dataset was first inspected for missing values, duplicate records, inconsistent formats, incorrect data types, and numerical outliers.

The following cleaning strategy will be applied:

1. Missing values will be handled using an appropriate strategy based on the nature and importance of each column.
2. Duplicate records will be identified and removed where they represent repeated records.
3. Inconsistent categorical formatting will be standardized.
4. Date columns will be converted into proper datetime format.
5. Numerical columns with incorrect data types will be converted to appropriate numeric types.
6. Outliers will be evaluated using the IQR method. Extreme values will not be removed automatically because some unusual values may represent legitimate transactions. The final treatment will be based on the context of each numerical variable.
7. The cleaned dataset will be saved as a new CSV file, while the original raw dataset will remain unchanged.

```python
# Detailed Missing Value Inspection

missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2),
    "Data Type": df.dtypes.astype(str)
})

missing_report = missing_report[missing_report["Missing Count"] > 0]

display(missing_report)
```

```python
# Remove Duplicate Records

duplicates_before = df.duplicated().sum()

print("Duplicate rows before removal:", duplicates_before)

df = df.drop_duplicates().copy()

duplicates_after = df.duplicated().sum()

print("Duplicate rows after removal:", duplicates_after)
print("Rows remaining:", len(df))
```

```python
# Standardize Text / Categorical Columns

text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
    df[column] = df[column].apply(
        lambda x: " ".join(x.split()) if isinstance(x, str) else x
    )

print("Text columns standardized:")
print(text_columns.tolist())
```

```python
# Identify and Convert Date Columns

date_columns = []

for column in df.columns:
    if "date" in column.lower():
        date_columns.append(column)

print("Date-related columns found:")
print(date_columns)

for column in date_columns:
    df[column] = pd.to_datetime(df[column], errors="coerce")

print("\nUpdated data types:")
print(df[date_columns].dtypes)
```

```python
# Check Numeric Columns and Their Data Types

print("=== NUMERIC COLUMNS ===")

numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    print(f"{column} → {df[column].dtype}")

print("\n=== NON-NUMERIC COLUMNS ===")

non_numeric_columns = df.select_dtypes(exclude=np.number).columns

for column in non_numeric_columns:
    print(f"{column} → {df[column].dtype}")
```

```python
# Detect Numeric Values Stored as Text

print("=== CHECKING OBJECT COLUMNS FOR NUMERIC VALUES ===")

object_columns = df.select_dtypes(include="object").columns

dtype_report = []

for column in object_columns:
    converted = pd.to_numeric(df[column], errors="coerce")
    
    original_non_null = df[column].notna().sum()
    numeric_values = converted.notna().sum()
    
    if original_non_null > 0:
        numeric_percentage = (numeric_values / original_non_null) * 100
    else:
        numeric_percentage = 0

    dtype_report.append({
        "Column": column,
        "Current Data Type": str(df[column].dtype),
        "Numeric-like Values (%)": round(numeric_percentage, 2)
    })

dtype_report = pd.DataFrame(dtype_report)

display(dtype_report)
```

```python
# Missing Values - Final Check Before Imputation

missing_before = pd.DataFrame({
    "Column": df.columns,
    "Missing Count": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().sum().values / len(df) * 100
    ).round(2),
    "Data Type": df.dtypes.astype(str).values
})

missing_before = missing_before[
    missing_before["Missing Count"] > 0
].sort_values("Missing Count", ascending=False)

display(missing_before)
```

## Missing Value Handling Strategy

Missing values were reviewed column-wise before applying any imputation.

- Numerical variables: Missing values will be handled using the median where appropriate because the median is less affected by extreme values.
- Categorical variables: Missing values will be handled using the mode where appropriate.
- Date variables: Missing dates will be reviewed separately and will not be replaced with arbitrary dates.
- Columns with a very high proportion of missing values will be evaluated based on their usefulness rather than blindly imputing them.

The selected treatment is intended to preserve as much valid information as possible while avoiding the introduction of artificial or misleading values.

```python
# Display Missing Value Report for Cleaning Decisions

display(missing_before)
```

```python
# Inspect values around columns containing missing data

columns_to_check = [
    "Discount Applied",
    "Item",
    "Price Per Unit",
    "Quantity",
    "Total Spent"
]

for column in columns_to_check:
    print(f"\n===== {column} =====")
    print(df[column].value_counts(dropna=False).head(15))
```

```python
# Inspect rows with missing numerical values

print("=== Missing Price Per Unit ===")
display(
    df[df["Price Per Unit"].isna()][
        ["Item", "Price Per Unit", "Quantity", "Total Spent"]
    ].head(10)
)

print("\n=== Missing Quantity ===")
display(
    df[df["Quantity"].isna()][
        ["Item", "Price Per Unit", "Quantity", "Total Spent"]
    ].head(10)
)

print("\n=== Missing Total Spent ===")
display(
    df[df["Total Spent"].isna()][
        ["Item", "Price Per Unit", "Quantity", "Total Spent"]
    ].head(10)
)
```

## Missing Value Treatment Decisions

The missing values were handled according to the meaning and relationships between the columns.

- **Discount Applied:** Missing values were replaced with `Unknown` because the existing True and False values are both common, and assuming either one could introduce bias.
- **Item:** Missing values were replaced with `Unknown` because the actual item cannot be reliably determined from the available data.
- **Price Per Unit:** Where Quantity and Total Spent were available, the missing price was calculated as `Total Spent / Quantity`. Any remaining missing prices were replaced using the median price.
- **Quantity:** Missing quantities were replaced using the median quantity because the original quantity cannot be reliably reconstructed when the required transaction information is unavailable.
- **Total Spent:** Where Price Per Unit and Quantity were available, missing values were calculated as `Price Per Unit × Quantity`. Any remaining missing values were replaced using the median Total Spent.

These decisions minimize unnecessary data loss while preserving the mathematical relationships between transaction variables.

```python
# Handle Missing Values

# 1. Categorical columns
df["Discount Applied"] = df["Discount Applied"].fillna("Unknown")
df["Item"] = df["Item"].fillna("Unknown")

# 2. Recover missing Price Per Unit where possible
price_mask = (
    df["Price Per Unit"].isna()
    & df["Quantity"].notna()
    & df["Total Spent"].notna()
    & (df["Quantity"] != 0)
)

df.loc[price_mask, "Price Per Unit"] = (
    df.loc[price_mask, "Total Spent"]
    / df.loc[price_mask, "Quantity"]
)

# Remaining missing Price Per Unit → median
df["Price Per Unit"] = df["Price Per Unit"].fillna(
    df["Price Per Unit"].median()
)

# 3. Missing Quantity → median
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

# 4. Recover missing Total Spent where possible
total_mask = (
    df["Total Spent"].isna()
    & df["Price Per Unit"].notna()
    & df["Quantity"].notna()
)

df.loc[total_mask, "Total Spent"] = (
    df.loc[total_mask, "Price Per Unit"]
    * df.loc[total_mask, "Quantity"]
)

# Remaining missing Total Spent → median
df["Total Spent"] = df["Total Spent"].fillna(
    df["Total Spent"].median()
)

print("Missing values handled successfully.")
```

```python
# Verify Missing Values After Cleaning

missing_after = df.isnull().sum()

print("=== MISSING VALUES AFTER CLEANING ===")
print(missing_after)

print("\nTotal missing values remaining:", missing_after.sum())
```

```python
# Final Outlier Report using IQR Method

numeric_columns = df.select_dtypes(include=np.number).columns

final_outlier_report = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    final_outlier_report.append({
        "Column": column,
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Outlier Count": outlier_count
    })

final_outlier_report = pd.DataFrame(final_outlier_report)

display(final_outlier_report)
```

```python
# Inspect Total Spent Outliers

Q1 = df["Total Spent"].quantile(0.25)
Q3 = df["Total Spent"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

total_spent_outliers = df[
    (df["Total Spent"] < lower_bound) |
    (df["Total Spent"] > upper_bound)
]

print("Number of Total Spent outliers:", len(total_spent_outliers))
print("Upper IQR limit:", round(upper_bound, 2))

display(
    total_spent_outliers[
        ["Item", "Price Per Unit", "Quantity", "Total Spent"]
    ].sort_values("Total Spent", ascending=False).head(10)
)
```

## Outlier Treatment Decision

The IQR method identified 60 outliers in the `Total Spent` column, with an upper IQR limit of 402.00.

The identified values were inspected and found to represent legitimate high-value transactions. For example, a transaction with a unit price of 41.00 and quantity of 10 has a total spent value of 410.00, which is mathematically valid.

Therefore, these outliers were **retained** rather than removed or capped. Removing them could result in the loss of valid transaction information.

No outliers were detected in `Price Per Unit` or `Quantity` using the IQR method.

```python
# Data Consistency Check
# Total Spent should be equal to Price Per Unit × Quantity

calculated_total = df["Price Per Unit"] * df["Quantity"]

difference = (df["Total Spent"] - calculated_total).abs()

print("Maximum difference:", difference.max())
print("Rows with inconsistent Total Spent:", (difference > 0.01).sum())

# Display inconsistent rows, if any
inconsistent_rows = df[difference > 0.01]

display(
    inconsistent_rows[
        ["Item", "Price Per Unit", "Quantity", "Total Spent"]
    ].head(10)
)
```
